```python
import numpy as np
import pandas as pd
from itertools import product

layer_height_range = [10]
layer_width_range = list(range(25,36))
extrusion_values = [3]

grid = list(product(extrusion_values, layer_width_range, layer_height_range))
search_grid = pd.DataFrame(grid, columns=['extrusion', 'layer_width', 'layer_height'])

search_grid = search_grid[['extrusion', 'layer_width', 'layer_height']]
search_grid['formula_speed'] = np.round(
    (search_grid['extrusion'] * 1e6) /
    (60 * ((search_grid['layer_width'] * search_grid['layer_height']) -
           (search_grid['layer_height']**2) +
           ((search_grid['layer_height']**2) * np.pi / 4)))
).astype(int)

display(search_grid)

search_grid.to_csv(f"data/experiment_ems_ext3.csv", index=False)
```
